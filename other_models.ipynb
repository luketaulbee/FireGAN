```python

# train_resnet6x_upgraded_patched.py
import os
import time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torch.optim.lr_scheduler import CosineAnnealingLR
import numpy as np
import h5py
import matplotlib.pyplot as plt
import pandas as pd
from tqdm import tqdm

# ----------------------------
# Configuration
# ----------------------------
class CFG:
    # raw base path (keep single slashes/backslashes consistent)
    raw_workpath = r'C:\Users\lhtau\OneDrive\Desktop\FireGAN_LukeTaulbee'
    workpath = os.path.normpath(raw_workpath)

    train_dir = os.path.normpath(os.path.join(workpath, 'Data', 'Train1'))
    test_dir = os.path.normpath(os.path.join(workpath, 'Data', 'Test1'))
    savedir = os.path.normpath(os.path.join(workpath, 'other_models', 'Saved_Models5'))
    image_out_dir = os.path.normpath(os.path.join(workpath, 'other_models', 'Images'))

    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    channels = 3  # BT04, BT05, AFI
    n_epochs = 5001
    batch_size = 4
    lr = 1e-4
    min_lr = 1e-5
    checkpoint_interval = 501
    use_perceptual = True  # optional perceptual loss

    # saving robustness
    save_retry_count = 3
    save_retry_delay = 1.0  # seconds

cfg = CFG()

# ensure dirs exist
os.makedirs(cfg.train_dir, exist_ok=True)
os.makedirs(cfg.test_dir, exist_ok=True)
os.makedirs(cfg.savedir, exist_ok=True)
os.makedirs(cfg.image_out_dir, exist_ok=True)

# ----------------------------
# Utilities
# ----------------------------
def calculate_rmse(pred, gt):
    return float(np.sqrt(np.mean((pred - gt)**2)))

def calculate_r2(pred, gt):
    ss_res = np.sum((gt - pred)**2)
    ss_tot = np.sum((gt - np.mean(gt))**2)
    return float(1 - ss_res / (ss_tot + 1e-8))

def calculate_psnr(pred, gt):
    mse = np.mean((pred - gt)**2)
    if mse == 0:
        return float('inf')
    PIXEL_MAX = np.max(gt) - np.min(gt)
    if PIXEL_MAX == 0:
        PIXEL_MAX = 1.0
    return float(20 * np.log10(PIXEL_MAX / np.sqrt(mse)))

def calculate_ssim(pred, gt):
    # naive SSIM approximation (keeps previous placeholder behavior)
    if np.max(gt) == np.min(gt):
        return 1.0 if np.allclose(pred, gt, atol=1e-6) else 0.0
    C1 = (0.01*np.max(gt))**2
    C2 = (0.03*np.max(gt))**2
    mu_x = np.mean(pred)
    mu_y = np.mean(gt)
    sigma_x = np.var(pred)
    sigma_y = np.var(gt)
    sigma_xy = np.mean((pred - mu_x)*(gt - mu_y))
    denom = (mu_x**2 + mu_y**2 + C1) * (sigma_x + sigma_y + C2)
    if denom == 0:
        return 0.0
    ssim = (2*mu_x*mu_y + C1)*(2*sigma_xy + C2) / denom
    return float(ssim)

def safe_save(obj, path, retry=3, delay=1.0):
    """
    Save atomically by writing to a temporary path then os.replace.
    Retry a few times in case of transient file-locks (e.g., OneDrive).
    """
    tmp = path + '.tmp'
    for attempt in range(1, retry+1):
        try:
            torch.save(obj, tmp)
            os.replace(tmp, path)
            return
        except Exception as e:
            if os.path.exists(tmp):
                try:
                    os.remove(tmp)
                except Exception:
                    pass
            if attempt == retry:
                raise
            time.sleep(delay)

# ----------------------------
# Dataset
# ----------------------------
class GOESVIIRSDataset(torch.utils.data.Dataset):
    def __init__(self, files):
        self.files = files

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        f = self.files[idx]
        with h5py.File(f, 'r') as h5f:
            lr = h5f['goes'][...]  # low-res: shape (C,h,w)
            hr = h5f['viirs'][...]  # high-res: shape (C,H,W)

        lr = lr.astype(np.float32)
        hr = hr.astype(np.float32)

        # -----------------------
        # Normalization functions
        # -----------------------
        def norm_viirs(data):
            data[0, :, :] = (data[0, :, :] - 220.0) / (350.0 - 220.0)
            data[1, :, :] = (data[1, :, :] - 220.0) / (320.0 - 220.0)
            data[2, :, :] = (data[2, :, :] - 0.0) / (8.0 - 0.0)
            return data

        def norm_goes(data):
            data[0, :, :] = (data[0, :, :] - 220.0) / (350.0 - 220.0)
            data[1, :, :] = (data[1, :, :] - 220.0) / (320.0 - 220.0)
            data[2, :, :] = (data[2, :, :] - 0.0) / (8.0 - 0.0)
            return data

        def denorm_viirs(data):
            data[0, :, :] = data[0, :, :] * (350.0 - 220.0) + 220.0
            data[1, :, :] = data[1, :, :] * (320.0 - 220.0) + 220.0
            data[2, :, :] = data[2, :, :] * (8.0 - 0.0) + 0.0
            return data

        def denorm_goes(data):
            data[0, :, :] = data[0, :, :] * (350.0 - 220.0) + 220.0
            data[1, :, :] = data[1, :, :] * (320.0 - 220.0) + 220.0
            data[2, :, :] = data[2, :, :] * (8.0 - 0.0) + 0.0
            return data

        # Apply normalization
        lr = norm_goes(lr)
        hr = norm_viirs(hr)

        # Convert to torch tensors
        lr = torch.from_numpy(lr).float()
        hr = torch.from_numpy(hr).float()
        return lr, hr, os.path.basename(f)

# ----------------------------
# Channel Attention Module
# ----------------------------
class CALayer(nn.Module):
    """Channel Attention"""
    def __init__(self, channel, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.conv_du = nn.Sequential(
            nn.Conv2d(channel, channel // reduction, 1, padding=0, bias=True),
            nn.ReLU(inplace=True),
            nn.Conv2d(channel // reduction, channel, 1, padding=0, bias=True),
            nn.Sigmoid()
        )
    def forward(self, x):
        y = self.avg_pool(x)
        y = self.conv_du(y)
        return x * y

# ----------------------------
# RRDB Block with Channel Attention
# ----------------------------
class ResidualDenseBlock_CA(nn.Module):
    def __init__(self, nf=256, gc=64):
        super().__init__()
        self.conv1 = nn.Conv2d(nf, gc, 3, 1, 1)
        self.conv2 = nn.Conv2d(nf+gc, gc, 3, 1, 1)
        self.conv3 = nn.Conv2d(nf+2*gc, gc, 3, 1, 1)
        self.conv4 = nn.Conv2d(nf+3*gc, gc, 3, 1, 1)
        self.conv5 = nn.Conv2d(nf+4*gc, nf, 3, 1, 1)
        self.lrelu = nn.LeakyReLU(0.2, inplace=True)
        self.ca = CALayer(nf)

    def forward(self, x):
        x1 = self.lrelu(self.conv1(x))
        x2 = self.lrelu(self.conv2(torch.cat([x,x1],1)))
        x3 = self.lrelu(self.conv3(torch.cat([x,x1,x2],1)))
        x4 = self.lrelu(self.conv4(torch.cat([x,x1,x2,x3],1)))
        x5 = self.conv5(torch.cat([x,x1,x2,x3,x4],1))
        x5 = self.ca(x5)
        return x + 0.2*x5

class RRDB_CA(nn.Module):
    def __init__(self, nf=256, gc=64):
        super().__init__()
        self.rdb1 = ResidualDenseBlock_CA(nf,gc)
        self.rdb2 = ResidualDenseBlock_CA(nf,gc)
        self.rdb3 = ResidualDenseBlock_CA(nf,gc)
    def forward(self,x):
        out = self.rdb1(x)
        out = self.rdb2(out)
        out = self.rdb3(out)
        return x + 0.2*out

# ----------------------------
# Upgraded ResNetSR6x
# ----------------------------
class ResNetSR6x_Upgraded(nn.Module):
    def __init__(self, channels=3, nf=256, n_blocks=48, gc=64):
        super().__init__()
        self.conv_first = nn.Conv2d(channels, nf, 3, 1, 1)
        self.blocks = nn.Sequential(*[RRDB_CA(nf=nf,gc=gc) for _ in range(n_blocks)])
        self.conv_trunk = nn.Conv2d(nf, nf, 3, 1, 1)
        self.lrelu = nn.LeakyReLU(0.2, inplace=True)
        self.upsample = nn.Sequential(
            nn.Conv2d(nf, nf*4, 3, 1, 1),
            nn.PixelShuffle(2),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(nf, nf*9, 3, 1, 1),
            nn.PixelShuffle(3),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(nf, channels, 3, 1, 1)
        )

    def forward(self, x):
        fea = self.lrelu(self.conv_first(x))
        trunk = self.conv_trunk(self.blocks(fea))
        fea = fea + trunk
        out = self.upsample(fea)
        return out

# ----------------------------
# Upgraded AE
# ----------------------------
class AutoencoderSR6x_Upgraded(nn.Module):
    def __init__(self, channels=3, nf=256, n_res_blocks=3):
        super().__init__()
        # encoder with extra conv + residual blocks
        encoder_layers = [nn.Conv2d(channels, nf, 3, 1, 1), nn.LeakyReLU(0.2, inplace=True)]
        for _ in range(3+n_res_blocks):  # 3 original + extra residual
            encoder_layers.append(nn.Conv2d(nf, nf, 3, 1, 1))
            encoder_layers.append(nn.LeakyReLU(0.2, inplace=True))
        self.encoder = nn.Sequential(*encoder_layers)

        # decoder with skip connections
        self.decoder_conv1 = nn.Conv2d(nf, nf*4, 3, 1, 1)
        self.pixelshuffle2 = nn.PixelShuffle(2)
        self.lrelu = nn.LeakyReLU(0.2, inplace=True)
        self.decoder_conv2 = nn.Conv2d(nf, nf*9, 3, 1, 1)
        self.pixelshuffle3 = nn.PixelShuffle(3)
        self.final_conv = nn.Conv2d(nf, channels, 3, 1, 1)

    def forward(self, x):
        fea = self.encoder(x)
        skip = fea  # skip connection
        fea = self.lrelu(self.decoder_conv1(fea))
        fea = self.pixelshuffle2(fea)
        fea = fea + skip  # fuse skip
        fea = self.lrelu(self.decoder_conv2(fea))
        fea = self.pixelshuffle3(fea)
        out = self.final_conv(fea)
        return out



# ----------------------------
# Loss functions
# ----------------------------
class PerceptualLoss(nn.Module):
    def __init__(self):
        super().__init__()
        try:
            vgg = torch.hub.load('pytorch/vision:v0.14.0','vgg16', pretrained=True)
            self.feature_extractor = nn.Sequential(*list(vgg.features)[:16]).eval()
            for p in self.feature_extractor.parameters(): p.requires_grad=False
        except Exception as e:
            # if hub fails (offline), gracefully disable perceptual
            print("Warning: VGG16 feature extractor failed to load:", e)
            self.feature_extractor = None
        self.l1 = nn.L1Loss()

    def forward(self, pred, gt):
        if self.feature_extractor is None:
            # fallback to L1 if vgg not available
            return self.l1(pred, gt)
        pred_feat = self.feature_extractor(pred)
        gt_feat = self.feature_extractor(gt)
        return self.l1(pred_feat, gt_feat)

# ----------------------------
# Training loop
# ----------------------------
def train():
    train_files = sorted([os.path.join(cfg.train_dir,f) for f in os.listdir(cfg.train_dir) if f.endswith('.h5')])
    if len(train_files) == 0:
        raise RuntimeError(f"No training files found in {cfg.train_dir} -- aborting.")
    ds_train = GOESVIIRSDataset(train_files)
    loader_train = DataLoader(ds_train, batch_size=cfg.batch_size, shuffle=True, num_workers=0,
                              pin_memory=True if cfg.device.startswith('cuda') else False)

    # Instantiate both models
    resnet = ResNetSR6x_(channels=cfg.channels, nf=128, n_blocks=32).to(cfg.device)
    ae = AutoencoderSR6x(channels=cfg.channels, nf=128).to(cfg.device)

    # Optimizers
    opt_resnet = torch.optim.Adam(resnet.parameters(), lr=cfg.lr)
    opt_ae = torch.optim.Adam(ae.parameters(), lr=cfg.lr)

    # LR schedulers
    sched_resnet = CosineAnnealingLR(opt_resnet, T_max=cfg.n_epochs, eta_min=cfg.min_lr)
    sched_ae = CosineAnnealingLR(opt_ae, T_max=cfg.n_epochs, eta_min=cfg.min_lr)

    # Loss
    criterion_l1 = nn.L1Loss()
    criterion_perc = PerceptualLoss().to(cfg.device) if cfg.use_perceptual else None

    scaler = torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())

    best_resnet_loss = float('inf')
    best_ae_loss = float('inf')

    history = {'resnet': [], 'ae': []}

    for epoch in range(1, cfg.n_epochs+1):
        resnet.train()
        ae.train()
        batch_loss_resnet = []
        batch_loss_ae = []

        pbar = tqdm(loader_train, desc=f'Epoch {epoch}')
        for lr_batch, hr_batch, _ in pbar:
            lr_batch = lr_batch.to(cfg.device)
            hr_batch = hr_batch.to(cfg.device)

            # --------------------
            # Train ResNet
            # --------------------
            opt_resnet.zero_grad()
            with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
                sr_out = resnet(lr_batch)
                loss_r = criterion_l1(sr_out, hr_batch)
                if cfg.use_perceptual and criterion_perc is not None:
                    loss_r += 0.2 * criterion_perc(sr_out, hr_batch)
            scaler.scale(loss_r).backward()
            scaler.step(opt_resnet)
            batch_loss_resnet.append(loss_r.item())

            # --------------------
            # Train AE
            # --------------------
            opt_ae.zero_grad()
            with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
                ae_out = ae(lr_batch)
                loss_ae = criterion_l1(ae_out, hr_batch)
                if cfg.use_perceptual and criterion_perc is not None:
                    loss_ae += 0.2 * criterion_perc(ae_out, hr_batch)
            scaler.scale(loss_ae).backward()
            scaler.step(opt_ae)
            scaler.update()
            batch_loss_ae.append(loss_ae.item())

            pbar.set_postfix({'ResNet_loss': loss_r.item(), 'AE_loss': loss_ae.item()})

        # Epoch metrics
        epoch_loss_r = float(np.mean(batch_loss_resnet))
        epoch_loss_ae = float(np.mean(batch_loss_ae))
        history['resnet'].append(epoch_loss_r)
        history['ae'].append(epoch_loss_ae)

        sched_resnet.step()
        sched_ae.step()

        # Save best ResNet
        if epoch_loss_r < best_resnet_loss:
            best_resnet_loss = epoch_loss_r
            safe_save(resnet.state_dict(), os.path.join(cfg.savedir, 'resnet_best.pth'),
                      retry=cfg.save_retry_count, delay=cfg.save_retry_delay)

        # Save best AE
        if epoch_loss_ae < best_ae_loss:
            best_ae_loss = epoch_loss_ae
            safe_save(ae.state_dict(), os.path.join(cfg.savedir, 'autoencoder_best.pth'),
                      retry=cfg.save_retry_count, delay=cfg.save_retry_delay)
        print(f"Epoch {epoch} complete. ResNet loss: {epoch_loss_r:.6f}, AE loss: {epoch_loss_ae:.6f}")

    # Save final weights
    safe_save(resnet.state_dict(), os.path.join(cfg.savedir,'resnet_final.pth'),
              retry=cfg.save_retry_count, delay=cfg.save_retry_delay)
    safe_save(ae.state_dict(), os.path.join(cfg.savedir,'autoencoder_final.pth'),
              retry=cfg.save_retry_count, delay=cfg.save_retry_delay)

    # Plot training curves
    plt.figure(figsize=(8,5))
    plt.plot(history['resnet'], label='ResNet train')
    plt.plot(history['ae'], label='AE train')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(cfg.image_out_dir,'training_loss_curve.png'), dpi=200)
    plt.show()



# ----------------------------
# Evaluation (complete with CSV + summary)
# ----------------------------
def safe_interp_upscale(tensor_lr, scale=6, mode='bicubic'):
    # expects tensor_lr shape (B,C,h,w) on CPU or device
    return F.interpolate(tensor_lr, scale_factor=scale, mode=mode, align_corners=False if mode in ('bicubic') else None)
def evaluate_test_set(model_paths: dict = None):
    """
    Evaluate AE (optional), ResNet (required), and bicubic interpolation on the test set.
    Produces:
      - metrics_csv (per-file rows)
      - test_metrics_summary_corrected.csv (per-channel & overall)
    """
    test_files = sorted([os.path.join(cfg.test_dir, f) for f in os.listdir(cfg.test_dir) if f.endswith('.h5')])
    if len(test_files) == 0:
        print("No test files found in:", cfg.test_dir)
        return None

    # model path defaults
    if model_paths is None:
        model_paths = {
            'AE': os.path.join(cfg.savedir, 'autoencoder_final.pth'),
            'ResNet': os.path.join(cfg.savedir, 'resnet_final.pth')
        }

    # Instantiate networks
    ae = None
    if os.path.exists(model_paths.get('AE', '')):
        try:
            from __main__ import AutoencoderSR6x
            ae = AutoencoderSR6x(channels=cfg.channels).to(cfg.device)
            ae.load_state_dict(torch.load(model_paths['AE'], map_location=cfg.device))
            ae.eval()
            print("Loaded AE from", model_paths['AE'])
        except Exception:
            print("AE not loaded. AE will be skipped.")
            ae = None

    rn = ResNetSR6x(channels=cfg.channels).to(cfg.device)
    if os.path.exists(model_paths['ResNet']):
        rn.load_state_dict(torch.load(model_paths['ResNet'], map_location=cfg.device))
        rn.eval()
    else:
        print("ResNet weights not found:", model_paths['ResNet'])
        return None

    interp_mode = 'bicubic'  # or 'bicubic', etc.
    models_to_eval = []
    if ae is not None:
        models_to_eval.append(('AE', ae))
    models_to_eval.append(('ResNet', rn))

    metrics_list = []
    loader_test = DataLoader(GOESVIIRSDataset(test_files), batch_size=1, shuffle=False, num_workers=0)

    with torch.no_grad():
        for lr_batch, hr_batch, fname in loader_test:
            lr_batch = lr_batch.to(cfg.device)
            hr_batch = hr_batch.to(cfg.device)

            ae_np = None
            if ae is not None:
                try:
                    ae_out = ae(lr_batch)
                    ae_np = ae_out.squeeze(0).cpu().numpy()
                except Exception as e:
                    print("AE forward failed for", fname, ":", e)
                    ae_np = None

            res_out = rn(lr_batch)
            res_np = res_out.squeeze(0).cpu().numpy()

            int_out = safe_interp_upscale(lr_batch, scale=6, mode=interp_mode)
            int_np = int_out.squeeze(0).cpu().numpy()

            gt_np = hr_batch.squeeze(0).cpu().numpy()

            def calculate_metrics_per_channel(pred_np, gt_np):
                d = {}
                for ch in range(cfg.channels):
                    p = pred_np[ch]
                    g = gt_np[ch]
                    d[f"CH{ch+1}_RMSE"] = calculate_rmse(p, g)
                    d[f"CH{ch+1}_R2"] = calculate_r2(p, g)
                    d[f"CH{ch+1}_PSNR"] = calculate_psnr(p, g)
                    d[f"CH{ch+1}_SSIM"] = calculate_ssim(p, g)
                return d

            metrics_dict = {"filename": fname[0]}
            for model_name, pred_np in zip(["AE", "ResNet", "Interp"], [ae_np, res_np, int_np]):
                if pred_np is None:
                    if model_name == "AE":
                        for ch in range(1, cfg.channels+1):
                            for m in ["RMSE","R2","PSNR","SSIM"]:
                                metrics_dict[f"{model_name}_CH{ch}_{m}"] = np.nan
                        continue
                ch_metrics = calculate_metrics_per_channel(pred_np, gt_np)
                metrics_dict.update({f"{model_name}_{k}": v for k, v in ch_metrics.items()})
            metrics_list.append(metrics_dict)

    # Save per-file metrics
    df = pd.DataFrame(metrics_list)
    metrics_csv = os.path.join(cfg.savedir, 'test_metrics_per_file.csv')
    df.to_csv(metrics_csv, index=False)
    print(f"\n Metrics saved to {metrics_csv}\n")

    # -----------------------------
    # Corrected per-channel & overall summary
    # -----------------------------
    summary_rows = []

    models = ["AE", "ResNet", "Interp"]
    channels = ["CH1", "CH2", "CH3"]
    metrics = ["RMSE", "R2", "PSNR", "SSIM"]

    print("\n================ CORRECTED PER-CHANNEL SUMMARY ================\n")
    for model in models:
        print(f"===== {model} =====")
        for ch in channels:
            ch_cols = [f"{model}_{ch}_{m}" for m in metrics if f"{model}_{ch}_{m}" in df.columns]
            if len(ch_cols) == 0:
                print(f"{ch}: No data")
                continue
            vals = df[ch_cols]
            means = vals.mean()
            stds = vals.std()
            summary_rows.append({
                "Model": model,
                "Channel": ch,
                **{f"{m}_mean": means[f"{model}_{ch}_{m}"] for m in metrics},
                **{f"{m}_std": stds[f"{model}_{ch}_{m}"] for m in metrics},
            })
            print(
                f"{ch}  "
                f"RMSE: {means[f'{model}_{ch}_RMSE']:.4f} ± {stds[f'{model}_{ch}_RMSE']:.4f}, "
                f"R²: {means[f'{model}_{ch}_R2']:.4f} ± {stds[f'{model}_{ch}_R2']:.4f}, "
                f"PSNR: {means[f'{model}_{ch}_PSNR']:.4f} ± {stds[f'{model}_{ch}_PSNR']:.4f}, "
                f"SSIM: {means[f'{model}_{ch}_SSIM']:.4f} ± {stds[f'{model}_{ch}_SSIM']:.4f}"
            )
        print()

    summary_df = pd.DataFrame(summary_rows)

    # Corrected overall summary (aggregate all channels)
    overall_metrics = {}
    for model in models:
        values = {m: [] for m in metrics}
        for ch in channels:
            for m in metrics:
                col_name = f"{model}_{ch}_{m}"
                if col_name in df.columns:
                    values[m].extend(df[col_name].dropna().tolist())
        overall_metrics[model] = values

    corrected_summary = {}
    for model, vals in overall_metrics.items():
        corrected_summary[model] = {}
        for m in metrics:
            arr = np.array(vals[m])
            corrected_summary[model][f"{m}_mean"] = arr.mean() if arr.size > 0 else np.nan
            corrected_summary[model][f"{m}_std"] = arr.std() if arr.size > 0 else np.nan

    print("\n================ CORRECTED OVERALL SUMMARY ================\n")
    for model, vals in corrected_summary.items():
        print(
            f"{model:<8}  "
            f"RMSE: {vals['RMSE_mean']:.4f} ± {vals['RMSE_std']:.4f}, "
            f"R²: {vals['R2_mean']:.4f} ± {vals['R2_std']:.4f}, "
            f"PSNR: {vals['PSNR_mean']:.4f} ± {vals['PSNR_std']:.4f}, "
            f"SSIM: {vals['SSIM_mean']:.4f} ± {vals['SSIM_std']:.4f}"
        )

    corrected_summary_path = os.path.join(cfg.workpath, "other_models", "test_metrics_summary_corrected.csv")
    pd.DataFrame.from_dict(corrected_summary, orient='index').to_csv(corrected_summary_path)
    print(f"\nCorrected overall summary saved to {corrected_summary_path}")

    return df, summary_df, corrected_summary


# ----------------------------
# Main
# ----------------------------
if __name__ == '__main__':
    # train()
    evaluate_test_set()



```

```python
# Full patched training & evaluation script (AE then ResNet sequential training)
# Patch: fixed plotting smoothing window bug (ensures smoothing window <= data length)
# - Applies previous fixes: class-name consistency, AMP scaler updates, AE skip upsample,
#   perceptual preprocessing, robust metric computations
# - Trains AE fully first (n_epochs_ae) then ResNet fully (n_epochs_resnet)
# - Produces enhanced visualizations and CSVs for both training phases
#
# Note: Change cfg.n_epochs_ae and cfg.n_epochs_resnet for your runs.
# Run a short smoke test with 1-2 epochs before long runs.

import os
import time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torch.optim.lr_scheduler import CosineAnnealingLR
import numpy as np
import h5py
import matplotlib.pyplot as plt
import pandas as pd
from tqdm import tqdm

# Try to import skimage SSIM for robust SSIM, fallback if missing
try:
    from skimage.metrics import structural_similarity as sk_ssim
    SKIMAGE_AVAILABLE = True
except Exception:
    SKIMAGE_AVAILABLE = False

# Optional tensorboard
try:
    from torch.utils.tensorboard import SummaryWriter
    TB_AVAILABLE = True
except Exception:
    TB_AVAILABLE = False

# ----------------------------
# Configuration
# ----------------------------
class CFG:
    raw_workpath = r'C:\Users\lhtau\OneDrive\Desktop\FireGAN_LukeTaulbee'
    workpath = os.path.normpath(raw_workpath)

    train_dir = os.path.normpath(os.path.join(workpath, 'Data', 'Train1'))
    test_dir = os.path.normpath(os.path.join(workpath, 'Data', 'Test1'))
    savedir = os.path.normpath(os.path.join(workpath, 'other_models', 'Saved_Models5'))
    image_out_dir = os.path.normpath(os.path.join(workpath, 'other_models', 'Images'))

    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    channels = 3  # BT04, BT05, AFI

    # Base epoch count (if not using separate)
    n_epochs = 50

    # Train sequential flags and per-model epoch counts
    train_sequential = True      # True => train AE fully then ResNet
    n_epochs_ae = 1001             # number of epochs to train AE (set as needed)
    n_epochs_resnet = 1001         # number of epochs to train ResNet (set as needed)

    batch_size = 4
    lr = 1e-4
    min_lr = 1e-5
    checkpoint_interval = 50
    use_perceptual = True  # optional perceptual loss

    # saving robustness
    save_retry_count = 3
    save_retry_delay = 1.0  # seconds

    # Model sizing defaults (tune if needed)
    resnet_nf = 128
    resnet_blocks = 8
    ae_nf = 128

    # Visualization smoothing windows
    smoothing_window_epochs = 3
    smoothing_window_batches = 200

cfg = CFG()

# ensure dirs
os.makedirs(cfg.train_dir, exist_ok=True)
os.makedirs(cfg.test_dir, exist_ok=True)
os.makedirs(cfg.savedir, exist_ok=True)
os.makedirs(cfg.image_out_dir, exist_ok=True)

# ----------------------------
# Utilities & denorm ranges
# ----------------------------
CHANNEL_MINS = np.array([220.0, 220.0, 0.0], dtype=np.float32)
CHANNEL_MAXS = np.array([350.0, 320.0, 8.0], dtype=np.float32)
CHANNEL_RANGES = CHANNEL_MAXS - CHANNEL_MINS

def calculate_rmse(pred, gt):
    return float(np.sqrt(np.mean((pred - gt)**2)))

def calculate_r2(pred, gt):
    ss_res = np.sum((gt - pred)**2)
    ss_tot = np.sum((gt - np.mean(gt))**2)
    return float(1 - ss_res / (ss_tot + 1e-8))

def calculate_psnr(pred, gt, data_range=1.0):
    mse = np.mean((pred - gt)**2)
    if mse == 0:
        return float('inf')
    return float(20 * np.log10(data_range / np.sqrt(mse)))

def calculate_ssim_single_channel(pred, gt, data_range=1.0):
    if SKIMAGE_AVAILABLE:
        try:
            return float(sk_ssim(pred, gt, data_range=float(data_range)))
        except Exception:
            pass
    C1 = (0.01 * data_range) ** 2
    C2 = (0.03 * data_range) ** 2
    mu_x = pred.mean()
    mu_y = gt.mean()
    sigma_x = pred.var()
    sigma_y = gt.var()
    sigma_xy = ((pred - mu_x) * (gt - mu_y)).mean()
    denom = (mu_x**2 + mu_y**2 + C1) * (sigma_x + sigma_y + C2)
    if denom == 0:
        return 0.0
    ssim = ((2 * mu_x * mu_y + C1) * (2 * sigma_xy + C2)) / denom
    return float(np.clip(ssim, -1.0, 1.0))

def safe_save(obj, path, retry=3, delay=1.0):
    tmp = path + '.tmp'
    for attempt in range(1, retry+1):
        try:
            torch.save(obj, tmp)
            os.replace(tmp, path)
            return
        except Exception:
            if os.path.exists(tmp):
                try:
                    os.remove(tmp)
                except Exception:
                    pass
            if attempt == retry:
                raise
            time.sleep(delay)

def denormalize_numpy(arr_norm):
    arr = arr_norm.copy().astype(np.float32)
    for c in range(arr.shape[0]):
        arr[c, :, :] = arr[c, :, :] * CHANNEL_RANGES[c] + CHANNEL_MINS[c]
    return arr

def denormalize_tensor(tensor):
    t = tensor.clone().detach().cpu().numpy()
    if t.ndim == 4:
        out = np.zeros_like(t)
        for b in range(t.shape[0]):
            out[b] = denormalize_numpy(t[b])
        return out
    else:
        return denormalize_numpy(t)

# ----------------------------
# Dataset
# ----------------------------
class GOESVIIRSDataset(torch.utils.data.Dataset):
    def __init__(self, files):
        self.files = files

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        f = self.files[idx]
        with h5py.File(f, 'r') as h5f:
            lr = h5f['goes'][...].astype(np.float32)
            hr = h5f['viirs'][...].astype(np.float32)

        def norm_viirs(data):
            data[0] = (data[0] - 220.0) / (350.0 - 220.0)
            data[1] = (data[1] - 220.0) / (320.0 - 220.0)
            data[2] = (data[2] - 0.0) / (8.0 - 0.0)
            return data

        def norm_goes(data):
            data[0] = (data[0] - 220.0) / (350.0 - 220.0)
            data[1] = (data[1] - 220.0) / (320.0 - 220.0)
            data[2] = (data[2] - 0.0) / (8.0 - 0.0)
            return data

        lr = norm_goes(lr)
        hr = norm_viirs(hr)

        lr = torch.from_numpy(lr).float()
        hr = torch.from_numpy(hr).float()
        return lr, hr, os.path.basename(f)

# ----------------------------
# Model components
# ----------------------------
class CALayer(nn.Module):
    def __init__(self, channel, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.conv_du = nn.Sequential(
            nn.Conv2d(channel, channel // reduction, 1, padding=0, bias=True),
            nn.ReLU(inplace=True),
            nn.Conv2d(channel // reduction, channel, 1, padding=0, bias=True),
            nn.Sigmoid()
        )
    def forward(self, x):
        y = self.avg_pool(x)
        y = self.conv_du(y)
        return x * y

class ResidualDenseBlock_CA(nn.Module):
    def __init__(self, nf=256, gc=64):
        super().__init__()
        self.conv1 = nn.Conv2d(nf, gc, 3, 1, 1)
        self.conv2 = nn.Conv2d(nf+gc, gc, 3, 1, 1)
        self.conv3 = nn.Conv2d(nf+2*gc, gc, 3, 1, 1)
        self.conv4 = nn.Conv2d(nf+3*gc, gc, 3, 1, 1)
        self.conv5 = nn.Conv2d(nf+4*gc, nf, 3, 1, 1)
        self.lrelu = nn.LeakyReLU(0.2, inplace=True)
        self.ca = CALayer(nf)

    def forward(self, x):
        x1 = self.lrelu(self.conv1(x))
        x2 = self.lrelu(self.conv2(torch.cat([x,x1],1)))
        x3 = self.lrelu(self.conv3(torch.cat([x,x1,x2],1)))
        x4 = self.lrelu(self.conv4(torch.cat([x,x1,x2,x3],1)))
        x5 = self.conv5(torch.cat([x,x1,x2,x3,x4],1))
        x5 = self.ca(x5)
        return x + 0.2*x5

class RRDB_CA(nn.Module):
    def __init__(self, nf=256, gc=64):
        super().__init__()
        self.rdb1 = ResidualDenseBlock_CA(nf,gc)
        self.rdb2 = ResidualDenseBlock_CA(nf,gc)
        self.rdb3 = ResidualDenseBlock_CA(nf,gc)
    def forward(self,x):
        out = self.rdb1(x)
        out = self.rdb2(out)
        out = self.rdb3(out)
        return x + 0.2*out

class ResNetSR6x_Upgraded(nn.Module):
    def __init__(self, channels=3, nf=256, n_blocks=48, gc=64):
        super().__init__()
        self.conv_first = nn.Conv2d(channels, nf, 3, 1, 1)
        self.blocks = nn.Sequential(*[RRDB_CA(nf=nf,gc=gc) for _ in range(n_blocks)])
        self.conv_trunk = nn.Conv2d(nf, nf, 3, 1, 1)
        self.lrelu = nn.LeakyReLU(0.2, inplace=True)
        self.upsample = nn.Sequential(
            nn.Conv2d(nf, nf*4, 3, 1, 1),
            nn.PixelShuffle(2),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(nf, nf*9, 3, 1, 1),
            nn.PixelShuffle(3),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(nf, channels, 3, 1, 1)
        )

    def forward(self, x):
        fea = self.lrelu(self.conv_first(x))
        trunk = self.conv_trunk(self.blocks(fea))
        fea = fea + trunk
        out = self.upsample(fea)
        return out

class AutoencoderSR6x_Upgraded(nn.Module):
    def __init__(self, channels=3, nf=256, n_res_blocks=3):
        super().__init__()
        encoder_layers = [nn.Conv2d(channels, nf, 3, 1, 1), nn.LeakyReLU(0.2, inplace=True)]
        for _ in range(3+n_res_blocks):
            encoder_layers.append(nn.Conv2d(nf, nf, 3, 1, 1))
            encoder_layers.append(nn.LeakyReLU(0.2, inplace=True))
        self.encoder = nn.Sequential(*encoder_layers)

        self.decoder_conv1 = nn.Conv2d(nf, nf*4, 3, 1, 1)
        self.pixelshuffle2 = nn.PixelShuffle(2)
        self.lrelu = nn.LeakyReLU(0.2, inplace=True)
        self.decoder_conv2 = nn.Conv2d(nf, nf*9, 3, 1, 1)
        self.pixelshuffle3 = nn.PixelShuffle(3)
        self.final_conv = nn.Conv2d(nf, channels, 3, 1, 1)

    def forward(self, x):
        fea = self.encoder(x)
        skip = fea
        fea = self.lrelu(self.decoder_conv1(fea))
        fea = self.pixelshuffle2(fea)
        skip_up = F.interpolate(skip, size=fea.shape[2:], mode='bilinear', align_corners=False)
        fea = fea + skip_up
        fea = self.lrelu(self.decoder_conv2(fea))
        fea = self.pixelshuffle3(fea)
        out = self.final_conv(fea)
        return out

# ----------------------------
# Perceptual Loss (with preprocessing)
# ----------------------------
imagenet_mean = torch.tensor([0.485, 0.456, 0.406], dtype=torch.float32).view(1,3,1,1)
imagenet_std = torch.tensor([0.229, 0.224, 0.225], dtype=torch.float32).view(1,3,1,1)

class PerceptualLoss(nn.Module):
    def __init__(self, device='cpu'):
        super().__init__()
        self.device = device
        try:
            vgg = torch.hub.load('pytorch/vision:v0.14.0','vgg16', pretrained=True)
            self.feature_extractor = nn.Sequential(*list(vgg.features)[:16]).eval().to(self.device)
            for p in self.feature_extractor.parameters(): p.requires_grad=False
        except Exception as e:
            print("Warning: VGG load failed:", e)
            self.feature_extractor = None
        self.l1 = nn.L1Loss()

    def forward(self, pred, gt):
        if self.feature_extractor is None:
            return self.l1(pred, gt)
        pred_den = denormalize_tensor(pred)
        gt_den = denormalize_tensor(gt)
        def to_t(x_np):
            return torch.from_numpy(x_np).float().to(self.device)
        pred_den_t = to_t(pred_den)
        gt_den_t = to_t(gt_den)
        ch_mins = torch.tensor(CHANNEL_MINS, dtype=torch.float32, device=self.device).view(1,-1,1,1)
        ch_ranges = torch.tensor(CHANNEL_RANGES, dtype=torch.float32, device=self.device).view(1,-1,1,1)
        pred01 = (pred_den_t - ch_mins) / (ch_ranges + 1e-8)
        gt01 = (gt_den_t - ch_mins) / (ch_ranges + 1e-8)
        pred01 = pred01.clamp(0.0,1.0)
        gt01 = gt01.clamp(0.0,1.0)
        pred_rs = F.interpolate(pred01, size=(224,224), mode='bilinear', align_corners=False)
        gt_rs = F.interpolate(gt01, size=(224,224), mode='bilinear', align_corners=False)
        mean = imagenet_mean.to(self.device)
        std = imagenet_std.to(self.device)
        pred_norm = (pred_rs - mean) / std
        gt_norm = (gt_rs - mean) / std
        pred_feat = self.feature_extractor(pred_norm)
        gt_feat = self.feature_extractor(gt_norm)
        return self.l1(pred_feat, gt_feat)

# ----------------------------
# Visualization helpers (patched to avoid smoothing length mismatch)
# ----------------------------
def moving_average(x, w):
    """
    moving average that ensures window size w is <= len(x) and handles empty input.
    Returns array with same length as x (or empty array if x empty).
    """
    x = np.asarray(x)
    if x.size == 0:
        return np.array([])
    w = int(max(1, w))
    if w > x.size:
        w = x.size
    if w <= 1:
        return x.copy()
    return np.convolve(x, np.ones(w)/w, mode='same')

def plot_training_history(history, out_dir, cfg, phase_name):
    os.makedirs(out_dir, exist_ok=True)
    batches = history.get(f'{phase_name}_batch', [])
    batch_epochs = history.get(f'{phase_name}_batch_epoch', [])
    epochs_list = history.get(f'{phase_name}_epoch', [])

    # Save CSVs
    df_batches = pd.DataFrame({
        'batch_index': np.arange(len(batches)) + 1,
        'epoch': batch_epochs,
        f'{phase_name}_batch_loss': batches
    })
    csv_path = os.path.join(out_dir, f'training_{phase_name}_batches.csv')
    df_batches.to_csv(csv_path, index=False)

    df_epochs = pd.DataFrame({
        'epoch': np.arange(1, len(epochs_list)+1),
        f'{phase_name}_epoch_loss': epochs_list
    })
    csv_epochs = os.path.join(out_dir, f'training_{phase_name}_epochs.csv')
    df_epochs.to_csv(csv_epochs, index=False)

    # Epoch-level plot
    epochs = df_epochs['epoch'].values if not df_epochs.empty else np.array([])
    e_loss = df_epochs[f'{phase_name}_epoch_loss'].values if not df_epochs.empty else np.array([])

    if e_loss.size > 0:
        win_e = max(1, int(cfg.smoothing_window_epochs))
        if win_e > e_loss.size:
            win_e = e_loss.size
        e_smooth = pd.Series(e_loss).rolling(window=win_e, min_periods=1, center=True).mean().values
        plt.figure(figsize=(8,4))
        plt.plot(epochs, e_loss, label=f'{phase_name} epoch loss', alpha=0.4, marker='o')
        plt.plot(epochs, e_smooth, label=f'{phase_name} smoothed (w={win_e})', linewidth=2)
        idx_min = int(np.nanargmin(e_loss))
        plt.scatter(epochs[idx_min], e_loss[idx_min], color='red', zorder=5)
        plt.annotate(f"best: {e_loss[idx_min]:.4f}\n(ep {epochs[idx_min]})",
                     xy=(epochs[idx_min], e_loss[idx_min]),
                     xytext=(epochs[idx_min], e_loss[idx_min]*1.1),
                     arrowprops=dict(arrowstyle='->', color='red'), fontsize=9)
        plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.title(f'{phase_name} epoch loss'); plt.grid(alpha=0.2)
        plt.legend()
        fname = os.path.join(out_dir, f'{phase_name}_epoch_loss.png')
        plt.tight_layout(); plt.savefig(fname, dpi=200); plt.close()
    else:
        fname = None

    # Batch-level plot
    batch_idx = df_batches['batch_index'].values if not df_batches.empty else np.array([])
    b_loss = df_batches[f'{phase_name}_batch_loss'].values if not df_batches.empty else np.array([])

    if b_loss.size > 0:
        win_b = max(1, int(cfg.smoothing_window_batches))
        if win_b > b_loss.size:
            win_b = b_loss.size
        b_smooth = moving_average(b_loss, win_b)
        plt.figure(figsize=(12,4))
        plt.plot(batch_idx, b_loss, alpha=0.15, label=f'{phase_name} batches')
        plt.plot(batch_idx, b_smooth, linewidth=1.6, label=f'{phase_name} batch-smooth (w={win_b})')
        for es in np.unique(batch_epochs):
            if len(df_batches[df_batches['epoch']==es]) > 0:
                start = int(df_batches[df_batches['epoch']==es]['batch_index'].iloc[0])
                plt.axvline(start, color='grey', linestyle='--', alpha=0.2)
        plt.xlabel('Batch index'); plt.ylabel('Loss'); plt.title(f'{phase_name} batch losses'); plt.legend(); plt.grid(alpha=0.2)
        fname2 = os.path.join(out_dir, f'{phase_name}_batch_loss.png')
        plt.tight_layout(); plt.savefig(fname2, dpi=200); plt.close()
    else:
        fname2 = None

    return {'batches_csv': csv_path, 'epochs_csv': csv_epochs, 'epoch_png': fname, 'batch_png': fname2}

# ----------------------------
# Train helper (single-phase)
# ----------------------------
def train_phase(model, model_name, n_epochs, ds_train, cfg, criterion_l1, criterion_perc, start_epoch=1):
    loader_train = DataLoader(ds_train, batch_size=cfg.batch_size, shuffle=True, num_workers=0,
                              pin_memory=True if cfg.device.startswith('cuda') else False)
    opt = torch.optim.Adam(model.parameters(), lr=cfg.lr)
    sched = CosineAnnealingLR(opt, T_max=n_epochs, eta_min=cfg.min_lr)
    scaler = torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())

    history = {f'{model_name}_epoch': [], f'{model_name}_batch': [], f'{model_name}_batch_epoch': []}
    best_loss = float('inf')

    tb_writer = SummaryWriter(log_dir=os.path.join(cfg.workpath, 'runs')) if TB_AVAILABLE else None
    global_batch_counter = 0

    for epoch in range(start_epoch, start_epoch + n_epochs):
        model.train()
        batch_losses = []
        pbar = tqdm(loader_train, desc=f'{model_name} Epoch {epoch}')
        for lr_batch, hr_batch, _ in pbar:
            lr_batch = lr_batch.to(cfg.device)
            hr_batch = hr_batch.to(cfg.device)
            opt.zero_grad()
            with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
                out = model(lr_batch)
                loss = criterion_l1(out, hr_batch)
                if cfg.use_perceptual and criterion_perc is not None:
                    loss = loss + 0.2 * criterion_perc(out, hr_batch)
            scaler.scale(loss).backward()
            scaler.step(opt)
            scaler.update()
            batch_losses.append(loss.item())
            history[f'{model_name}_batch'].append(loss.item())
            global_batch_counter += 1
            history[f'{model_name}_batch_epoch'].append(epoch)
            if tb_writer is not None:
                tb_writer.add_scalar(f'batch/{model_name}_loss', loss.item(), global_batch_counter)
            pbar.set_postfix({f'{model_name}_loss': loss.item()})
        epoch_loss = float(np.mean(batch_losses)) if len(batch_losses)>0 else float('nan')
        history[f'{model_name}_epoch'].append(epoch_loss)
        if tb_writer is not None:
            tb_writer.add_scalar(f'epoch/{model_name}_loss', epoch_loss, epoch)
        if epoch_loss < best_loss:
            best_loss = epoch_loss
            safe_save(model.state_dict(), os.path.join(cfg.savedir, f'{model_name.lower()}_best.pth'),
                      retry=cfg.save_retry_count, delay=cfg.save_retry_delay)
        sched.step()
        print(f"{model_name} Epoch {epoch} complete. {model_name} loss: {epoch_loss:.6f}")

    if tb_writer is not None:
        tb_writer.close()
    safe_save(model.state_dict(), os.path.join(cfg.savedir, f'{model_name.lower()}_final.pth'),
              retry=cfg.save_retry_count, delay=cfg.save_retry_delay)
    return history

# ----------------------------
# Full training (AE then ResNet)
# ----------------------------
def train():
    train_files = sorted([os.path.join(cfg.train_dir,f) for f in os.listdir(cfg.train_dir) if f.endswith('.h5')])
    if len(train_files) == 0:
        raise RuntimeError(f"No training files found in {cfg.train_dir} -- aborting.")
    ds_train = GOESVIIRSDataset(train_files)

    # Instantiate models
    ae = AutoencoderSR6x_Upgraded(channels=cfg.channels, nf=cfg.ae_nf).to(cfg.device)
    resnet = ResNetSR6x_Upgraded(channels=cfg.channels, nf=cfg.resnet_nf, n_blocks=cfg.resnet_blocks).to(cfg.device)

    criterion_l1 = nn.L1Loss()
    criterion_perc = PerceptualLoss(device=cfg.device).to(cfg.device) if cfg.use_perceptual else None

    all_histories = {}

    if cfg.train_sequential:
        # Train AE fully first
        print("Starting AE training for", cfg.n_epochs_ae, "epochs")
        hist_ae = train_phase(ae, 'AE', cfg.n_epochs_ae, ds_train, cfg, criterion_l1, criterion_perc, start_epoch=1)
        all_histories.update(hist_ae)
        viz_ae = plot_training_history({'AE_epoch': hist_ae['AE_epoch'], 'AE_batch': hist_ae['AE_batch'], 'AE_batch_epoch': hist_ae['AE_batch_epoch']},
                                      cfg.image_out_dir, cfg, 'AE')
        print("AE training done. Plots saved:", viz_ae)

        # Train ResNet fully next
        print("Starting ResNet training for", cfg.n_epochs_resnet, "epochs")
        hist_res = train_phase(resnet, 'ResNet', cfg.n_epochs_resnet, ds_train, cfg, criterion_l1, criterion_perc, start_epoch=1)
        all_histories.update(hist_res)
        viz_rn = plot_training_history({'ResNet_epoch': hist_res['ResNet_epoch'], 'ResNet_batch': hist_res['ResNet_batch'], 'ResNet_batch_epoch': hist_res['ResNet_batch_epoch']},
                                       cfg.image_out_dir, cfg, 'ResNet')
        print("ResNet training done. Plots saved:", viz_rn)
    else:
        raise NotImplementedError("Non-sequential training not implemented in this script. Set cfg.train_sequential=True.")

    return all_histories

# ----------------------------
# Evaluation with both denorm and normalized metrics
# ----------------------------
def safe_interp_upscale(tensor_lr, scale=6, mode='bicubic'):
    return F.interpolate(tensor_lr, scale_factor=scale, mode=mode, align_corners=False)

def evaluate_test_set(model_paths: dict = None):
    test_files = sorted([os.path.join(cfg.test_dir, f) for f in os.listdir(cfg.test_dir) if f.endswith('.h5')])
    if len(test_files) == 0:
        print("No test files found in:", cfg.test_dir)
        return None

    if model_paths is None:
        model_paths = {
            'AE': os.path.join(cfg.savedir, 'ae_final.pth'),
            'ResNet': os.path.join(cfg.savedir, 'resnet_final.pth')
        }

    ae = None
    if os.path.exists(model_paths.get('AE', '')):
        try:
            ae = AutoencoderSR6x_Upgraded(channels=cfg.channels, nf=cfg.ae_nf).to(cfg.device)
            ae.load_state_dict(torch.load(model_paths['AE'], map_location=cfg.device))
            ae.eval()
            print("Loaded AE from", model_paths['AE'])
        except Exception as e:
            print("AE not loaded. AE will be skipped. Error:", e)
            ae = None

    rn = ResNetSR6x_Upgraded(channels=cfg.channels, nf=cfg.resnet_nf, n_blocks=cfg.resnet_blocks).to(cfg.device)
    if os.path.exists(model_paths['ResNet']):
        try:
            rn.load_state_dict(torch.load(model_paths['ResNet'], map_location=cfg.device))
            rn.eval()
            print("Loaded ResNet from", model_paths['ResNet'])
        except Exception as e:
            print("Failed to load ResNet weights:", e)
            return None
    else:
        print("ResNet weights not found:", model_paths['ResNet'])
        return None

    interp_mode = 'bicubic'
    loader_test = DataLoader(GOESVIIRSDataset(test_files), batch_size=1, shuffle=False, num_workers=0)

    metrics_list = []
    with torch.no_grad():
        for lr_batch, hr_batch, fname in tqdm(loader_test, desc='Evaluating'):
            lr_batch = lr_batch.to(cfg.device)
            hr_batch = hr_batch.to(cfg.device)

            ae_np = None
            if ae is not None:
                try:
                    ae_out = ae(lr_batch)
                    ae_np = ae_out.squeeze(0).cpu().numpy()
                except Exception as e:
                    print("AE forward failed for", fname, ":", e)
                    ae_np = None

            res_out = rn(lr_batch)
            res_np = res_out.squeeze(0).cpu().numpy()

            int_out = safe_interp_upscale(lr_batch, scale=6, mode=interp_mode)
            int_np = int_out.squeeze(0).cpu().numpy()

            gt_np = hr_batch.squeeze(0).cpu().numpy()

            # denorm arrays
            def denorm(arr_norm):
                out = arr_norm.copy().astype(np.float32)
                for c in range(out.shape[0]):
                    out[c] = out[c] * CHANNEL_RANGES[c] + CHANNEL_MINS[c]
                return out

            gt_den = denorm(gt_np)
            res_den = denorm(res_np)
            int_den = denorm(int_np)
            ae_den = denorm(ae_np) if ae_np is not None else None

            # normalized arrays (0-1)
            def to_norm(den):
                out = den.copy().astype(np.float32)
                for c in range(out.shape[0]):
                    out[c] = (out[c] - CHANNEL_MINS[c]) / (CHANNEL_RANGES[c] + 1e-8)
                return out

            gt_norm = to_norm(gt_den)
            res_norm = to_norm(res_den)
            int_norm = to_norm(int_den)
            ae_norm = to_norm(ae_den) if ae_den is not None else None

            metrics_dict = {'filename': fname[0]}
            for model_name, pred_den, pred_norm in zip(['AE','ResNet','Interp'], [ae_den, res_den, int_den], [ae_norm, res_norm, int_norm]):
                if pred_den is None:
                    if model_name == 'AE':
                        for ch in range(cfg.channels):
                            for m in ['RMSE','R2','PSNR','SSIM']:
                                metrics_dict[f"{model_name}_CH{ch+1}_{m}"] = np.nan
                                metrics_dict[f"{model_name}_CH{ch+1}_{m}_NORM"] = np.nan
                        continue
                for ch in range(cfg.channels):
                    p_den = pred_den[ch]
                    g_den = gt_den[ch]
                    p_norm = pred_norm[ch]
                    g_norm = gt_norm[ch]

                    rmse_den = calculate_rmse(p_den, g_den)
                    r2_den = calculate_r2(p_den, g_den)
                    psnr_den = calculate_psnr(p_den, g_den, data_range=float(CHANNEL_RANGES[ch]))
                    ssim_den = calculate_ssim_single_channel(p_den, g_den, data_range=float(CHANNEL_RANGES[ch]))

                    rmse_norm = calculate_rmse(p_norm, g_norm)
                    r2_norm = calculate_r2(p_norm, g_norm)
                    psnr_norm = calculate_psnr(p_norm, g_norm, data_range=1.0)
                    ssim_norm = calculate_ssim_single_channel(p_norm, g_norm, data_range=1.0)

                    metrics_dict[f"{model_name}_CH{ch+1}_RMSE"] = rmse_den
                    metrics_dict[f"{model_name}_CH{ch+1}_R2"] = r2_den
                    metrics_dict[f"{model_name}_CH{ch+1}_PSNR"] = psnr_den
                    metrics_dict[f"{model_name}_CH{ch+1}_SSIM"] = ssim_den

                    metrics_dict[f"{model_name}_CH{ch+1}_RMSE_NORM"] = rmse_norm
                    metrics_dict[f"{model_name}_CH{ch+1}_R2_NORM"] = r2_norm
                    metrics_dict[f"{model_name}_CH{ch+1}_PSNR_NORM"] = psnr_norm
                    metrics_dict[f"{model_name}_CH{ch+1}_SSIM_NORM"] = ssim_norm

            metrics_list.append(metrics_dict)

    df = pd.DataFrame(metrics_list)
    metrics_csv = os.path.join(cfg.savedir, 'test_metrics_per_file_with_norm.csv')
    df.to_csv(metrics_csv, index=False)
    print("Saved metrics (denorm + normalized) to:", metrics_csv)

    # Summary printing (denorm per-channel & overall)
    def print_summary_from_df(df):
        print("\n================ CORRECTED PER-CHANNEL SUMMARY (DENORM) ================\n")
        models = ['AE','ResNet','Interp']
        channels = [f"CH{i}" for i in range(1, cfg.channels+1)]
        metrics = ['RMSE','R2','PSNR','SSIM']
        for model in models:
            print(f"===== {model} =====")
            for ch in channels:
                ch_cols = [f"{model}_{ch}_{m}" for m in metrics if f"{model}_{ch}_{m}" in df.columns]
                if len(ch_cols) == 0:
                    print(f"{ch}: No data")
                    continue
                vals = df[ch_cols]
                means = vals.mean()
                stds = vals.std()
                print(
                    f"{ch}  "
                    f"RMSE: {means[f'{model}_{ch}_RMSE']:.4f} ± {stds[f'{model}_{ch}_RMSE']:.4f}, "
                    f"R²: {means[f'{model}_{ch}_R2']:.4f} ± {stds[f'{model}_{ch}_R2']:.4f}, "
                    f"PSNR: {means[f'{model}_{ch}_PSNR']:.4f} ± {stds[f'{model}_{ch}_PSNR']:.4f}, "
                    f"SSIM: {means[f'{model}_{ch}_SSIM']:.4f} ± {stds[f'{model}_{ch}_SSIM']:.4f}"
                )
            print()
        # overall
        print("\n================ CORRECTED OVERALL SUMMARY (DENORM) ================\n")
        for model in models:
            vals_agg = {m: [] for m in ['RMSE','R2','PSNR','SSIM']}
            for ch in channels:
                for m in vals_agg.keys():
                    col = f"{model}_{ch}_{m}"
                    if col in df.columns:
                        vals_agg[m].extend(df[col].dropna().tolist())
            print(f"{model:<8}  RMSE: {np.nanmean(vals_agg['RMSE']):.4f} ± {np.nanstd(vals_agg['RMSE']):.4f}, "
                  f"R²: {np.nanmean(vals_agg['R2']):.4f} ± {np.nanstd(vals_agg['R2']):.4f}, "
                  f"PSNR: {np.nanmean(vals_agg['PSNR']):.4f} ± {np.nanstd(vals_agg['PSNR']):.4f}, "
                  f"SSIM: {np.nanmean(vals_agg['SSIM']):.4f} ± {np.nanstd(vals_agg['SSIM']):.4f}")
    print_summary_from_df(df)

    return df

# ----------------------------
# Main
# ----------------------------
if __name__ == '__main__':
    # Run sequential training: AE then ResNet
    # histories = train()
    evaluate_test_set()
```
